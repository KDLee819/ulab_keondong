## 1 Functions
**Orbital Periods and Kepler's Third Law** \
Kepler's Third Law states that the square of the orbital period $T$ (in years) of a planet is proportional to the cube of the semi-major axis $a$ (in astronomical units, AU):

$$
T^2 = a^3
$$

Write a **function** that calculates the orbital period $T$ given the semi-major axis $a$. (So, you can assume that $T = \sqrt{a^3}$.)

```python
# Your code here (WITHOUT NUMPY! or any other modules/libraries/packages)

def calculate_period(a):
    '''
    Function name: calculate_period
    Input: semi-major axis a, in units of AU
    Output: orbital period T, in units of years
    
    '''
    # Calculate period using the given Kepler's third law, and name it orbital_period
    orbital_period = a ** 1.5

    # return this as the output of the function 
    return orbital_period
```

Call your function with the following values: 
* `planet_a` $ = 0.5\,\text{AU}$
* `planet_b` $ = 1.0\,\text{AU}$
* `planet_c` $ = 1.5\,\text{AU}$

```python
# Your code here
# Semi-major axis for each planet, in units of AU
a_a = 0.5
a_b = 1.0
a_c = 1.5

# Calculate period for each planets
T_a = calculate_period(a_a)
T_b = calculate_period(a_b)
T_c = calculate_period(a_c)

# Print the results
print(f"Planet A has a semi-major axis of {a_a} AU and have period of about {round(T_a, 2)} years.")
print(f"Planet B has a semi-major axis of {a_b} AU and have period of about {round(T_b, 2)} years.")
print(f"Planet C has a semi-major axis of {a_c} AU and have period of about {round(T_c, 2)} years.")
```

## 2 Indexing + Conditionals

You are provided with a list of exoplanet radii (in Earth radii) and their equilibrium temperatures (in Kelvin):

`radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]` \
`temperatures = [300, 400, 350, 280, 500, 450]`

Write a loop to filter and print the radii of planets with temperatures below 350 K.

```python
# Your code here

# Write down given list
radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]
temperatures = [300, 400, 350, 280, 500, 450]

# Initialize an empty list where the filtered values of planet radii would be stored
filtered_radii = []

# Construct a for-loop
for i in range(len(temperatures)):
    # Run over temperature values

    # append radii which has the same index with temperatures below 350K
    if temperatures[i] < 350:
        filtered_radii.append(radii[i])

# Print the modified empty list
print(filtered_radii)
```

Use list slicing to extract and print the radii of the first three planets.

```python
# Your code here

# Slice radii as instructed
# No index means from the beginning, and includes up to index before the written index
sliced_radii = radii[:3]

print(sliced_radii)
```

Turn the radii and temperatures data into a dictionary. 

```python
# Your code here

# Define lists
radii = [1.2, 0.8, 1.5, 2.0, 1.0, 0.6]
temperatures = [300, 400, 350, 280, 500, 450]

# Initialize empty list
combined_dict = {}

# Iterate through the lists using while-loop
i = 0
while i < 6:
    combined_dict[f"Planet {i}"] = [radii[i], temperatures[i]]
    i += 1

# Print out the result
print(combined_dict)
```

Give each of the planets in your dictionary a name. 

```python
# Your code here

# This is a list of Planet names
planet_name_list = ["Mars", "Venus", "Jupiter", "Saturn", "Earth", "Mercury"]

# Raplace each keys by names with a for-loop, and a 'pop' method
for i in range(len(planet_name_list)):
    combined_dict[planet_name_list[i]] = combined_dict.pop(f"Planet {i}") # Deletes old keys and replace it with the new name

# Print out the new dictionary
print(combined_dict)
```

Create a new dictionary of planets with radii between 1.0 and 2.0 Earth radii (inclusive).

```python
# Your code here

# Initialize empty dictionary where filtered values would be stored
filtered_dict = {}

# Make lists with each keys and values
name_list = list(combined_dict.keys())
size_list = list(combined_dict.values())

# Construct a for-loop that filters required radii range
for i in range(len(name_list)):
    if size_list[i][0] <= 2.0 and size_list[i][0] >= 1.0:
        # Add to empty list if radius is between 1.0 and 2.0 (inclusive!!)
        filtered_dict[name_list[i]] = size_list[i]

# Print out the result
print(filtered_dict)
```

Use a for loop to print the value on the second row, third column. 

```python
# Your code here

# Iterate through arrays using for-loop
# Double for-loop because we are doing with multi-dimensional array
for i in range(len(md_array)): # First for-loop accounts row (index 0 = 1st row, etc.)
    
    for j in range(len(md_array[0])): # Second for-loop accounts column (index 2 = 3rd column, etc.)
        
        if i == 1 and j == 2:
            print(md_array[i][j]) # Print the conrresponding value if it is in the second row, third column
            
        else:
            continue # Do not print any other values
```

## 4 Plotting
Create a scatter plot of the following semi-major axes (x-axis) vs. orbital periods (y-axis). Label the axes, add a legend and add a title: "Orbital Periods of Exoplanets."

`semi_major_axes = [0.5, 1.0, 2.0, 1.5, 1.2, 0.8]`

```python
# Your code here
# Import necessary modules
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Re-define values using np.array
semi_major_axes = np.array([0.5, 1.0, 2.0, 1.5, 1.2, 0.8])
orbital_periods = np.array(semi_major_axes ** 1.5)

plt.scatter(semi_major_axes, orbital_periods, label="a vs T data points")
plt.xlabel("Semi-major axes, $a$ (AU)")
plt.ylabel("Orbital periods, $T$ (years)")
plt.legend()
plt.title("Orbital Periods of Exoplanets")
plt.show()
```

Create a bar plot of planet radii vs. their equilibrium temperatures. Label the axes, add a legend and add a title: "Equilibrium Temperatures vs. Radii."

```python
# Your code here
# Re-define values using np.array
planet_radii = np.array(radii)
equi_temperatures = np.array(temperatures)

plt.plot(planet_radii, equi_temperatures, label="radii vs temperature", color="green")
plt.xlabel("Planet radii, $R$ (Earth radius)")
plt.ylabel("Equilibrium temperature, $T$ (°K)")
plt.legend()
plt.title("Equilibrium Temperatures vs. Radii")
plt.show()
```

Put the scatter plot and bar plot on the same figure (horizonatally). 

```python
# Your code here

# Design subplots with 1 row and 2 column
fig, axs = plt.subplots(1, 2, figsize=(12, 4))

axs[0].scatter(semi_major_axes, orbital_periods, label="a vs T data points")
axs[0].set_title("Orbital Periods of Exoplanets")
axs[0].set_xlabel("Semi-major axes, $a$ (AU)")
axs[0].set_ylabel("Orbital periods, $T$ (years)")
axs[0].legend()

axs[1].plot(planet_radii, equi_temperatures, label="radii vs temperature", color="green")
axs[1].set_title("Equilibrium Temperatures vs. Radii")
axs[1].set_xlabel("Planet radii, $R$ (Earth radius)")
axs[1].set_ylabel("Equilibrium temperature, $T$ (°K)")
axs[1].legend()

plt.show()
```
